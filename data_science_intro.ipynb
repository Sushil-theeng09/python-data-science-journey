```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Hours_Studied": [1, 2, 3, 4, 5],
    "Marks": [45, 52, 65, 78, 92]
})

data
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

sns.regplot(data=data, x="Hours_Studied", y="Marks")

plt.title("Study Hours vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.show()
```
